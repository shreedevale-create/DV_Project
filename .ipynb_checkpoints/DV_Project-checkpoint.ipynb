```python
movies.info()                    # column types, missing values
movies.describe()                # numeric summaries
movies["release_date"] = pd.to_datetime(movies["release_date"])
movies["year"] = movies["release_date"].dt.year
movies["genres"].head(10)        # note: multiple genres in one string!
```






## 3. Data preparation 

```python
# dates → year and decade
movies["release_date"] = pd.to_datetime(movies["release_date"])
movies["year"] = movies["release_date"].dt.year
movies["decade"] = (movies["year"] // 10) * 10

# money subset: only films with real budget AND revenue figures
money = movies[(movies["budget"] > 0) & (movies["revenue"] > 0)].copy()
money["profit"] = money["revenue"] - money["budget"]
money["roi"] = money["profit"] / money["budget"]
print("money subset:", money.shape)

# genre view: one row per movie-genre combination
genre_df = movies.dropna(subset=["genres"]).copy()
genre_df["genres"] = genre_df["genres"].str.split(", ")
genre_df = genre_df.explode("genres")
print("genre rows:", genre_df.shape)
```

## Q1 — Do bigger budgets earn better ratings, or just more revenue?

```python
print("budget vs rating: ", money["budget"].corr(money["vote_average"]).round(2))
print("budget vs revenue:", money["budget"].corr(money["revenue"]).round(2))
```

```python
import plotly.express as px

fig = px.scatter(
    money, x="budget", y="revenue",
    color="vote_average",
    log_x=True, log_y=True,          # money spans huge ranges → log scales
    hover_name="title",
    color_continuous_scale="Viridis", # CVD-safe palette
    labels={"budget": "Budget ($, log)", "revenue": "Revenue ($, log)", "vote_average": "Rating"},
    title="Bigger budgets buy box office, not quality",
)
fig.show()
```

## Q2: Which genres deliver the best ROI, and has that changed by decade?

```python
# genre-exploded version of the money subset
money_genre = money.dropna(subset=["genres"]).copy()
money_genre["genres"] = money_genre["genres"].str.split(", ")
money_genre = money_genre.explode("genres")

# median ROI per genre per decade (median, not mean — one 100x hit would distort everything)
roi = (money_genre[money_genre["decade"] >= 1970]
       .groupby(["decade", "genres"])["roi"]
       .agg(["median", "count"])
       .reset_index())
roi = roi[roi["count"] >= 20]     # only genre-decades with enough films to trust
roi.sort_values("median", ascending=False).head(10)
```

```python
pivot = roi.pivot(index="genres", columns="decade", values="median")

fig = px.imshow(
    pivot,
    color_continuous_scale="Viridis",
    labels={"x": "Decade", "y": "", "color": "Median ROI"},
    title="PLACEHOLDER — write the takeaway after you see it",
    aspect="auto",
)
fig.show()
```

## Q3: Are flops and blockbusters concentrated in certain genres? 

```python
# order genres by median profit, keep genres with enough films
counts = money_genre["genres"].value_counts()
big_genres = counts[counts >= 100].index
sub = money_genre[money_genre["genres"].isin(big_genres)]

order = sub.groupby("genres")["profit"].median().sort_values().index

fig = px.box(
    sub, x="profit", y="genres",
    category_orders={"genres": list(order)},
    labels={"profit": "Profit ($)", "genres": ""},
    title="PLACEHOLDER — takeaway",
)
fig.update_xaxes(range=[-2e8, 1e9])   # zoom past extreme outliers; note it in your insight
fig.show()
```
